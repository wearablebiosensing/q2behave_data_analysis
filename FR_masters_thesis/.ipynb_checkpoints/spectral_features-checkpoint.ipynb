```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import scipy as sc
from scipy.signal import find_peaks
import scipy as sc
from scipy.stats import skew, kurtosis
from scipy.signal import welch
import os

def calculate_sampling_frequency(dataframe):
    frequencies = dataframe['timestamp_str_seconds'].value_counts()
    average_frequency = sum(frequencies) / len(frequencies)

    return average_frequency


def plot_psd_my(df):
    selected_columns = df.iloc[:, 4:10]
    print("selected_columns: ",selected_columns.columns)
    N = len(selected_columns)
    T = 1/calculate_sampling_frequency(df)
    num_rows = 2
    num_cols = 3
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 8))
    plt.subplots_adjust(hspace=0.35, wspace=0.65)
    
    for k in range(num_rows):
        for j in range(num_cols):
            column_index = k * num_cols + j
            if column_index >= selected_columns.shape[1]:
                # If there are no more columns to plot, break out of the loop
                break
            data_no_dc = selected_columns.iloc[:, column_index].values
    
            ax = axes[k, j]  # Get the current axis from the 'axes' array
            ax.psd(data_no_dc, NFFT=N, Fs=1/T, color='blue', label='PSD') # alpha=0.7
    
            ax.grid()
            ax.set_title(f"{selected_columns.columns[column_index]} - PSD, no DC")
    
    plt.suptitle('')    
    plt.show() 
```

```python
def plot_psd(df):
    selected_columns = df.iloc[:, 4:10]
    N = len(selected_columns)
    T = 1/calculate_sampling_frequency(df)
    num_rows = 2
    num_cols = 3
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 8))
    plt.subplots_adjust(hspace=0.5, wspace=0.35)  # Increase the spacing between rows
    feature_sets = []  # List to store feature sets for each axis of each sensor

    for k in range(num_rows):
        for j in range(num_cols):
            column_index = k * num_cols + j
            if column_index >= selected_columns.shape[1]:
                # If there are no more columns to plot, break out of the loop
                break
            data_no_dc = selected_columns.iloc[:, column_index].values
            
            ax = axes[k, j]  # Get the current axis from the 'axes' array
            freqs, psd = welch(data_no_dc, fs=1/T, nperseg=N/2)
            psd = np.abs(psd)  # Take absolute value to ensure positive values
            
            # Find peaks using scipy's find_peaks
            peak_indices, _ = find_peaks(psd)
            peaks_freq = freqs[peak_indices]
            peaks_psd = psd[peak_indices]
            peak_diffs_freq = np.diff(peaks_freq)
            print("peak_diffs_freq: ",peak_diffs_freq)
            feature_max =max(psd)
            feature_min = min(psd)
            feature_skewness = skew(psd)
            feature_kurtosis = kurtosis(psd)
            feature_mean = sum(psd)/len(psd)
            feature_sum_freq_diff = sum(peak_diffs_freq)
            feature_average_power = sum(peaks_psd)/len(peaks_psd)
            max_index = np.argmax(psd)
            freq_at_max = freqs[max_index]

            # Store the calculated features in a dictionary
            feature_set = {
                f"feature_max_{selected_columns.columns[column_index]}": feature_max,
                f"feature_min_{selected_columns.columns[column_index]}": feature_min,
                f"feature_skewness_{selected_columns.columns[column_index]}": feature_skewness,
                f"feature_kurtosis_{selected_columns.columns[column_index]}": feature_kurtosis,
                f"feature_mean_{selected_columns.columns[column_index]}": feature_mean,
                f"feature_sum_freq_diff_{selected_columns.columns[column_index]}": feature_sum_freq_diff,
                f"feature_average_power_{selected_columns.columns[column_index]}": feature_average_power
            }
            feature_sets.append(feature_set)
            
            ax.plot(freqs, psd, color='blue', label='PSD')
            ax.plot(peaks_freq, peaks_psd, 'ro', markersize=5, label='Peaks')  # Plot peaks
            ax.set_yscale('log')  # Set y-axis to logarithmic scale
            ax.grid()
            ax.set_title(f"{selected_columns.columns[column_index]} - PSD, no DC" + "\n"+ "Mean: "+  str(sum(psd)/len(psd)) + 
                         "\n"+ "Maximum: "+ str(max(psd)) + "\n"
                         + "Minimum: "+ str(min(psd))  +"\n"+
                        "Feq at max: "+ str(freq_at_max) + 
                        "\n" + "Skewness: " + str(skew(psd)) + "\n" + "Kurtosis: " + str(kurtosis(psd)))
    plt.suptitle('')
    plt.show()
    # Create a dataframe from the list of feature sets
    feature_df = pd.DataFrame(feature_sets)
    return feature_df
```

```python

def calculate_features(axis_data, T, N, axis_name):
    freqs, psd = welch(axis_data, fs=1/T, nperseg=N/2)
    psd = np.abs(psd)  # Take absolute value to ensure positive values

    # Find peaks using scipy's find_peaks
    peak_indices, _ = find_peaks(psd)
    peaks_freq = freqs[peak_indices]
    peaks_psd = psd[peak_indices]
    peak_diffs_freq = np.diff(peaks_freq)

    feature_max = max(psd)
    feature_min = min(psd)
    feature_skewness = skew(psd)
    feature_kurtosis = kurtosis(psd)
    feature_mean = sum(psd) / len(psd)
    feature_sum_freq_diff = sum(peak_diffs_freq)
    feature_average_power = sum(peaks_psd) / len(peaks_psd)
    feature_num_peaks = len(peaks_psd)
    return {
        f"{axis_name}_max_power_psd": feature_max,
        f"{axis_name}_min_power_psd": feature_min,
        f"{axis_name}_skewness_power_psd": feature_skewness,
        f"{axis_name}_kurtosis_power_psd": feature_kurtosis,
        f"{axis_name}_mean_power_psd": feature_mean,
        f"{axis_name}_sum_freq_diff_power_psd": feature_sum_freq_diff,
        f"{axis_name}_average_power_power_psd": feature_average_power,
        f"{axis_name}feature_num_peaks_power_psd": feature_num_peaks
    }

def extract_psd_features(df):
    selected_columns = df.iloc[:, 4:10]
    N = len(selected_columns)
    T = 1 / calculate_sampling_frequency(df)
    feature_set = {}  # Dictionary to store feature sets
    
    for axis in selected_columns.columns:
        axis_data = selected_columns[axis].values
        features = calculate_features(axis_data, T, N, axis)
        feature_set.update(features)
    
    # Convert the dictionary to a dataframe with a single row
    feature_df = pd.DataFrame([feature_set])
    return feature_df


```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import scipy as sc
from scipy.signal import find_peaks
#from scipy.stats import skew, kurtosis
#from sklearn.preprocessing import MinMaxScaler

import os

def calculate_sampling_frequency(dataframe):
    frequencies = dataframe['timestamp_str_seconds'].value_counts()
    average_frequency = sum(frequencies) / len(frequencies)

    return average_frequency



def dominant_frequency(data, sampling_rate):
    data_no_dc = data# - data.mean()
    fft_result = np.fft.fft(data_no_dc)
    freqs = np.fft.fftfreq(len(fft_result), 1/sampling_rate)
    dominant_freq = np.abs(freqs[np.argmax(np.abs(fft_result))])
    return dominant_freq

def maximum_amplitude(data, sampling_rate,):
    data_no_dc = data #- data.mean()
    fft_result = np.fft.fft(data_no_dc)
    #freqs = np.fft.fftfreq(len(fft_result), 1/sampling_rate)
    # Calculate maximum amplitude for each column
    max_amplitude = np.max(np.abs(fft_result))
    return max_amplitude


df = pd.read_csv('C:\\Users\\fried\\Documents\\URI\\ELE_591\\syncedfiles\\BlobsOFSegments\\P10_Left_A2_Standing_up_while_working__1.csv')
#df = pd.read_csv('C:\\Users\\fried\\Documents\\URI\\ELE_591\\syncedfiles\\BlobsOFSegments\\P10_Left_A2_NoHyp_2.csv')



selected_columns = df.iloc[:, 4:10]
N = len(selected_columns)
fs = calculate_sampling_frequency(df)
T = 1/fs#calculate_sampling_frequency(df)


num_rows = 2
num_cols = 3
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 8))
plt.subplots_adjust(hspace=0.35, wspace=0.35)

for k in range(num_rows):
    for j in range(num_cols):
        column_index = k * num_cols + j
        if column_index >= selected_columns.shape[1]:
            # If there are no more columns to plot, break out of the loop
            break
        
        data_no_dc = selected_columns.iloc[:, column_index].values     
        #data_no_dc = selected_columns.iloc[:, column_index].values - selected_columns.iloc[:, column_index].values.mean()

        yf = sc.fft.fft(data_no_dc)
        xf = sc.fft.fftfreq(N, T)[:N//2]

        ax = axes[k, j]  
        ax.plot(xf, 2.0/N * np.abs(yf[0:N//2]))
        ax.grid()
        ax.set_title(selected_columns.columns[column_index])
        
       
        dominant_freq = dominant_frequency(data_no_dc, fs)
        max_amplitude = maximum_amplitude(data_no_dc, fs)

        title_text = f"{selected_columns.columns[column_index]} - PSD\n"
        title_text += f"Dominant Frequency: {dominant_freq:.2f} Hz\n"
        title_text += f"Max Amplitude: {max_amplitude:.2f}"

        ax.set_title(title_text)


plt.suptitle('')
#plt.suptitle('P10_Left_A2_NoHyp_2')


plt.show()
```
