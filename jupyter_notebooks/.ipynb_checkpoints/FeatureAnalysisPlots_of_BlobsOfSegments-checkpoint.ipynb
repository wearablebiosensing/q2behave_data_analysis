```python
"""
Helper function to plot the PSD while taking it out.
"""
def my_plot_psd(df,Fs):
    selected_columns = df[[' Accel_X',' Accel_Y',' Accel_Z',' Gyro_X',' Gyro_Y',' Gyro_Z']]
    N = len(selected_columns)
    print("nperseg: ",N,N/2)
    T = 1/Fs #@calculate_sampling_frequency(df)
    num_rows = 2
    num_cols = 3
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 8))
    plt.subplots_adjust(hspace=0.5, wspace=0.35)  # Increase the spacing between rows
    feature_sets = []  # List to store feature sets for each axis of each sensor

    for k in range(num_rows):
        for j in range(num_cols):
            column_index = k * num_cols + j
            if column_index >= selected_columns.shape[1]:
                # If there are no more columns to plot, break out of the loop
                break
            data_no_dc = selected_columns.iloc[:, column_index].values
            
            ax = axes[k, j]  # Get the current axis from the 'axes' array
            freqs, psd = welch(data_no_dc, fs=1/T) # nperseg=int(N/2)
            psd = np.abs(psd)  # Take absolute value to ensure positive values
            
            # Find peaks using scipy's find_peaks
            peak_indices, _ = find_peaks(psd)
            peaks_freq = freqs[peak_indices]
            peaks_psd = psd[peak_indices]
            peak_diffs_freq = np.diff(peaks_freq)
            ax.plot(freqs, psd, color='blue', label='PSD')
            ax.plot(peaks_freq, peaks_psd, 'ro', markersize=5, label='Peaks')  # Plot peaks
            ax.set_yscale('log')  # Set y-axis to logarithmic scale
            ax.grid()
            ax.set_title(f"{selected_columns.columns[column_index]} - PSD, no DC")
    plt.suptitle('')
    plt.show()
```

```python
def average_sampling_rate(df):
    sr_dict = df["NewClock"].value_counts().to_dict()
    #print("average_sampling_rate():/ sr_dict: ",sr_dict)
    # Step 1: Calculate the sum of all values
    total_sum = sum(sr_dict.values())
    # Step 2: Count the number of values
    num_values = len(sr_dict)
    if num_values !=0:
        #print("num_values: ",num_values)
        #print("total_sum: ",total_sum)
        # Step 3: Calculate the average
        average_sr = total_sum / num_values
        return average_sr
        
def my_fft(df_column,Fs):
    Acc_max_np_original = df_column.values
    # Calculate the mean of the array
    mean_value = np.mean(Acc_max_np_original)
    # Subtract the mean from each value in the array to remove DC bias
    Acc_max_np = Acc_max_np_original - mean_value

    #print("Acc_max_np type(Acc_max_np) ",type(Acc_max_np))
    N = len(Acc_max_np)
    signal_fft = scipy.fftpack.fft(Acc_max_np)
    signal_amp = 2 / N* np.abs(signal_fft)
    signal_freq = np.abs(fftpack.fftfreq(N, Fs/1000))

    return signal_fft,signal_freq, signal_amp
# https://nehajirafe.medium.com/using-fft-to-analyse-and-cleanse-time-series-data-d0c793bb82e3
def top_two_freq(fft, signal_amp,seconds):
    #print("signal_amp: ",signal_amp)
    #Calculate Frequency Magnitude
    # magnitudes = abs(fft[np.where(signal_freq > 0)])
  #  print("top_two_freq():/ signal_amp: ",signal_amp)
   # print("top_two_freq():/ np.argpartition(magnitudes, -2)[-2:]): ",np.argpartition(signal_amp, -2)[-2:])
    #Get index of top 2 frequencies
    peak_frequency = np.sort((np.argpartition(signal_amp, -2)[-2:])/seconds)
    print("top_two_freq():/ Top 2 peak frequencies: ",peak_frequency)
    return peak_frequency #[###,##] arrary of length 2. consiosts fo top 2 dominant frequiencies.


def butter_lowpass_filter(data, cutoff, fs, order):
    print("Cutoff freq " + str(cutoff))
    nyq = 0.5 * fs # Nyquist Frequency
    normal_cutoff = cutoff / nyq
    # Get the filter coefficients 
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    y = filtfilt(b, a,data)
    return y
def duration_of_data(subset_df):
    begin_ts =  subset_df["NewClock"].values[0]
    end_ts = subset_df["NewClock"].values[len(subset_df["NewClock"].values)-1]
    print("duration_of_data():/ begin_ts: ",begin_ts)
    print("duration_of_data():/ end_ts: ",end_ts)
    # Convert timestamp strings to datetime objects
    time_format = "%H:%M:%S"
    datetime1 = datetime.strptime(begin_ts, time_format)
    datetime2 = datetime.strptime(end_ts, time_format)
    
    # Calculate the difference in seconds
    difference_seconds = (datetime2 - datetime1).total_seconds()
    
    print("Number of seconds between the two timestamps:", difference_seconds)
    return difference_seconds

```

```python
#full_activities_folder
def windowed_features(df, window_size, overlap):
    max_behavior_list = []
    num_samples = len(df)
    window_counter = 0
    start = 0
    dfs = []
    Fs = average_sampling_rate(df)
    # max_values = df.max(axis=1)
    max_A = df[[' Accel_X', ' Accel_Y',' Accel_Z']].max(axis=1)
    max_G = df[[' Gyro_X', ' Gyro_Y', ' Gyro_Z']].max(axis=1)
    
    df["Acc_max"] = max_A
    df["Gry_max"] = max_G

    ### Process the windowed data feature to get all the features #######.
    moving_average_filter_window_size = 5
    # moving_avereaged_df = apply_moving_avg(df,moving_average_filter_window_size,Fs)
    while start < num_samples:
        end = min(start + window_size, num_samples)
        window_data = df.iloc[int(start):int(end)]
        #window_data = df.iloc[start:end]
        #print("windowed_features()===========:/  window_data",window_data["NewClock"])
        print("Window Number: ",window_counter )
        # Get The behaviour with maximum count.
        behavior_counts = window_data['HyperactiveBehaviourType'].value_counts()
        max_behavior = behavior_counts.idxmax() if not behavior_counts.empty else None
        print("behaviour: ",max_behavior)
        fig_move_chair = filtering_plots_raw_plotly(window_data)
        fig_move_chair.show()
        max_behavior_list.append(max_behavior)
        window_counter+=1
        # yield start, end, window_data, max_behavior
        start += window_size - overlap
```

```python
 # Function to extract the time part from the timestamp
def extract_time(timestamp):
    return timestamp.split(':')[1:]

def get_new_int_timestamps(sorted_data):
    np_ts_array_list = []
    counter = 1
    for i in sorted_data:
        print("counter ===============",counter)
        print(i, ",",sorted_data[i])
        int_ts = np.full(sorted_data[i], counter)
        for i in int_ts:
            np_ts_array_list.append(i)
        print("int_ts: ",int_ts)
        counter +=1
    return np_ts_array_list
    
# Define a mapping function
def map_values(value):
    if pd.isna(value):
        return 0
    elif 'chair' in value:
        return 1
    elif 'Standing' in value:
        return 2
    elif "Manipulating" in value:
        return 3
    elif "Twirling" in value:
        return 4
    elif "Drumming" in value:
        return 5
    elif "Finger" in value:
        return 6
    elif "NoHyp" in value:
        return 0     
    else:
        return -1  # for values not matching any criteria

# Define a mapping function
def map_values_handedness(value):
    if pd.isna(value):
        return 0
    elif 'Left' in value:
        return "Left"
    elif 'Right' in value:
        return "Right"
    else:
        return -1  # for values not matching any criteria
```

```python
def plot_spectograms(df, filename):
    average_sr1 = average_sampling_rate(df)
    min_signal_length = min(df.apply(len))
    # Calculate a reasonable value for nperseg
    nperseg1 = int(2 * average_sr1) 
    print("plot_spectograms(): nperseg1 or window size: ", nperseg1)
    print("plot_spectograms(): Overlap: ", nperseg1 // 3)
    max_behavior_list = iterate_over_windows(df, window_size=nperseg1, overlap=nperseg1 // 3)
    # Setting fonts and styles.
    title_font = 20
    axis_font = 10
    desired_font_size = 20  # Adjust as needed.
    font = {'family': 'Sans'}
    plt.rc('font', **font)
    plt.rc('xtick', labelsize=30) 
    plt.rc('ytick', labelsize=30)  
    # rows, cols 
    fig, axs = plt.subplots(4, 3, figsize=(30, 10)) # width, heights.
    plt.subplots_adjust(hspace=0.5)  
    # Iterate over accelerometer axes
    for i, axis in enumerate([' Accel_X', ' Accel_Y', ' Accel_Z']):
        # Compute the spectrogram for the current axis
        f, t, Sxx = signal.spectrogram(df[axis], average_sr1, nperseg=nperseg1, noverlap=nperseg1 // 3)
        # print("Sxx : ",Sxx)
        energry = np.sum(Sxx, axis=0)
        energry_withlog = np.sum(10 * np.log10(Sxx), axis=0)
        
        # First Row: Spectrogram
        axs[0, i].set_title(f'Accelerometer {axis}' + "without log", fontsize=title_font)
        axs[0, i].set_xlabel('Time', fontsize=desired_font_size)
        axs[0, i].set_ylabel('Frequency (Hz)', fontsize=desired_font_size)
        axs[0, i].set_ylim(0, 4)
        specgram1 = axs[0, i].pcolormesh(t, f, Sxx, shading='gouraud',cmap="binary")
        # axs[0, i].plot(max_behavior_list[:-1],linewidth=4, label="Behavior types")

        # First Row: Spectrogram
        axs[1, i].set_title(f'Accelerometer {axis}' + "with log", fontsize=title_font)
        axs[1, i].set_xlabel('Time with log', fontsize=desired_font_size)
        axs[1, i].set_ylabel('Frequency (Hz)', fontsize=desired_font_size)
        axs[1, i].set_ylim(0, 4)
        specgram2 = axs[1, i].pcolormesh(t, f, 10 * np.log10(Sxx), shading='gouraud',cmap="binary")
        # # Filtered Data Moving Average.
        axs[2, i].set_title("Energry plots without log scale", fontsize=title_font)
        axs[2, i].plot(max_behavior_list[:-1],linewidth=4, label="Behavior types")
        axs[2, i].plot(t,energry, color="blue", linewidth=4, label="Energy without log")
        axs[3, i].set_title("Energry plots with log", fontsize=title_font)
        axs[3, i].plot(t,energry_withlog, color="blue", linewidth=4, label="Energy with log")        
    plt.tight_layout()
    plt.show()
```

```python

def read_synced_files(folder_path):
    dfs = []
    file_list = os.listdir(folder_path)
    
    for filename in file_list:
        if filename.endswith('.csv'):
            file_path = os.path.join(folder_path, filename)
            df = pd.read_csv(file_path)
            dfs.append(df)
    dfs_concat = pd.concat(dfs)
    return dfs_concat
    
    # Now, 'dfs' is a list of DataFrames, each representing a CSV file in the folder
```

```python
def average_sampling_rate(df):
    sr_dict = df["NewClock"].value_counts().to_dict()
    # Step 1: Calculate the sum of all values
    total_sum = sum(sr_dict.values())
    # Step 2: Count the number of values
    num_values = len(sr_dict)
    if num_values !=0:
        #print("num_values: ",num_values)
        #print("total_sum: ",total_sum)
        # Step 3: Calculate the average
        average_sr = total_sum / num_values
        return average_sr
    else:
        return "na"
```

```python
# Convert the elapsed_time strings to seconds
def to_seconds(time_string):
    minutes, seconds = map(int, time_string.split(":")[1:])
    return minutes * 60 + seconds
```

```python

def iterate_over_windows(df, window_size, overlap):
    """
    Iterate over a pandas DataFrame with a specified window size and overlap.

    Parameters:
        df (pandas DataFrame): The DataFrame to iterate over.
        window_size (int): The size of each window (number of samples).
        overlap (int): The number of samples to overlap between windows.
    """
    max_behavior_list = []
    num_samples = len(df)
    start = 0
    while start < num_samples:
        end = min(start + window_size, num_samples)
        window_data = df.iloc[start:end]
        behavior_counts = window_data['BehaviorCode'].value_counts()
        max_behavior = behavior_counts.idxmax() if not behavior_counts.empty else None
        max_behavior_list.append(max_behavior)
        # yield start, end, window_data, max_behavior
        start += window_size - overlap
    print("iterate_over_windows(): max_behavior_list: ",len(max_behavior_list))
    return max_behavior_list


```

```python
## Add codes to string behaviour types.##
def pre_datacleaning(row):
    ################################################
    # Filter out specific behavior types.
    ################################################
    if "Standing" in row['HyperactiveBehaviourType']:
        return 1
    elif "chair" in row['HyperactiveBehaviourType']:
        return 2
    elif "Manipulating" in row['HyperactiveBehaviourType']:
        return 3
    elif "Twirling" in row['HyperactiveBehaviourType']:
        return 4
    elif "Drumming" in row['HyperactiveBehaviourType']:
        return 5
    elif "Finger" in row['HyperactiveBehaviourType']:
        return 6
    elif "NoHyp" in row['HyperactiveBehaviourType']:
        return 0
    else:
        return "None"  # Or whatever default behavior you want
```

##  Gross Angle Estimations

```python
################################################################################################################
# Calculate Joint Angles using dot product.
################################################################################################################

def dot_product(a, b):
    return np.dot(a, b)

def magnitude(vector):
    return np.linalg.norm(vector)

def angle_between_vectors(a, b):
    dot = dot_product(a, b)
    mag_a = magnitude(a)
    mag_b = magnitude(b)
    if mag_a == 0 or mag_b == 0:
        return None  # Prevent division by zero
    cosine_angle = dot / (mag_a * mag_b)
    return math.acos(cosine_angle)
```

```python
# Function to calculate angles between accelerometer and gyroscope axes
def calculate_angles(row):
    accelerometer = row[[' Accel_X',' Accel_Z',' Accel_Z']].values
    gyroscope = row[[' Gyro_X', ' Gyro_Y', ' Gyro_Z']].values
   # print("Rows accelerometer:", accelerometer)
    #print("Rows gyroscope :", gyroscope)
    # Check if any of the accelerometer or gyroscope values are None
    # if None in accelerometer or None in gyroscope:
    #     print("Skipping row with None values:", row)
    #     return pd.Series([None], index=['angle_acc_gyro_x'])
    angle = angle_between_vectors(accelerometer, gyroscope)
    if angle == None:
        print("Skipping row with None values:", row)
        return pd.Series([None], index=['angle_acc_gyro_x'])
    
    angle_acc_gyro_x = math.degrees(angle)
    

    return pd.Series([angle_acc_gyro_x], index=['angle_acc_gyro_x'])
```

```python
# SPECTOGRAM PLOTS UP NEXT.....
# using MATPLOT LIB we are able to print all the graphs on the screen itself.
def joint_angle_estimations(folder_path):
    dfs = []
    file_list = os.listdir(folder_path)
    for filename in file_list:
        if filename.endswith('.csv'):
            file_path = os.path.join(folder_path, filename)
            print("filename: ",filename)
            print("===========================================================================================================================================")
            df = pd.read_csv(file_path,low_memory=False)
            df = df[df["Hyperactive_Restless"] != -100]
            print("HyperactiveBehaviourType: ",df["HyperactiveBehaviourType"].unique())
            df['HyperactiveBehaviourType'] = df['HyperactiveBehaviourType'].fillna('NaN').astype(str)
            df["BehaviorCode"] = df.apply(pre_datacleaning, axis=1)
            print("df.columns: ",df.columns)
            print("ACC ,y,z",df.iloc[:, 4:7])
            print("GRY,y,z",df.iloc[:, 7:10])
            df['Accel_RMS'] = np.sqrt(np.sum(df.iloc[:, 4:7] ** 2, axis = 1))
            df['Gry_RMS'] = np.sqrt(np.sum(df.iloc[:, 7:10] ** 2, axis = 1))

            # Apply the function to each row of the DataFrame
            angles_df = df.apply(calculate_angles, axis=1)
            print("angles_df: ",angles_df)
            
            title_font = 20
            axis_font = 20
            bins=40
            font = {'family' : 'Sans'}
            plt.rc('font', **font)
            plt.rc('xtick', labelsize=10) 
            plt.rc('ytick', labelsize=10) 
            # rows,cols
            fig, axs  = plt.subplots(2, 2,figsize=(15, 5))
            plt.subplots_adjust(hspace = 0.5) # top=0.8-> for top title spacing.

            # First Row
            axs[0,0].set_title('Estimated Angles in Degrees' +filename, fontsize=title_font)
            axs[0,0].plot(angles_df,color="red",label  = "Estimated Angles")
           
            # Add some space between subplots
            plt.tight_layout()
            plt.show()
```

### Feature Extraction

```python
# features to analyse
#mean
def feature_mean(feature_name,accel_gyro_data):
    mean = accel_gyro_data.mean()
    mean_transposed_df = mean.to_frame().T
    mean_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name]
    return mean_transposed_df
    
#std
def feature_std(feature_name,accel_gyro_data):
    std = accel_gyro_data.std()
    std_transposed_df = std.to_frame().T
    std_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name]
    return std_transposed_df

#kurtosis
def feature_kurtosis(feature_name,accel_gyro_data):
    kurtosis = accel_gyro_data.kurtosis()
    kurtosis_transposed_df = kurtosis.to_frame().T
    kurtosis_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_"+feature_name]
    return kurtosis_transposed_df

#median
def feature_median(feature_name,accel_gyro_data):
    median = accel_gyro_data.median()
    median_transposed_df = median.to_frame().T
    median_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" + feature_name]
    return median_transposed_df

#skewness
def feature_skewness(feature_name,accel_gyro_data):
    skewness = accel_gyro_data.skew()
    skewness_transposed_df = skewness.to_frame().T
    skewness_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" + feature_name]
    return skewness_transposed_df
```

```python
# vector distance mean
def feature_sum(feature_name,accel_gyro_data):
    sum_ = accel_gyro_data.sum()
    sum_transposed_df = sum_.to_frame().T
    sum_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name]
    #["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name ,"Magno_X_" +feature_name, "Magno_Y_" +feature_name ,"Magno_Z_" +feature_name, "Accel_Vecdis_" +feature_name, "Gyro_Vecdis_" +feature_name ,"Magno_Vecdis_" +feature_name]
    return sum_transposed_df

def calculate_zero_crossing_rate(accel_data):
    accel_data = (accel_data - np.mean(accel_data)) / np.std(accel_data)
        # Calculate Zerocrossing rate.
    zero_crossings = np.nonzero(np.diff(np.signbit(accel_data)))[0]
    zcr = len(zero_crossings) / (2.0 * len(accel_data))
    return zcr

def feature_zero_cross(feature_name,accel_gyro_data):
    zero_crossing_rates = {}
    for column in accel_gyro_data.columns:
        zcr = calculate_zero_crossing_rate(accel_gyro_data[column])
        zero_crossing_rates[column] = zcr
    zcr_series = pd.Series(zero_crossing_rates)
    zcr_transposed_df = zcr_series.to_frame().T
    zcr_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name]
    #["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name ,"Magno_X_" +feature_name, "Magno_Y_" +feature_name ,"Magno_Z_" +feature_name]
    return zcr_transposed_df


# vector distance mean
def diff_sum(feature_name,accel_gyro_data):
    sum_ = accel_gyro_data.diff().sum()
    sum_transposed_df = sum_.to_frame().T
    sum_transposed_df.columns = ["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name]
    #["Accel_X_"+feature_name ,"Accel_Y_" +feature_name ,"Accel_Z_" +feature_name,"Gyro_X_" +feature_name, "Gyro_Y_" + feature_name ,"Gyro_Z_" +feature_name ,"Magno_X_" +feature_name, "Magno_Y_" +feature_name ,"Magno_Z_" +feature_name, "Accel_Vecdis_" +feature_name, "Gyro_Vecdis_" +feature_name ,"Magno_Vecdis_" +feature_name]
    return sum_transposed_df
```

```python
# create table 
def get_concated_features(accel_gyro_data):    
    df1 = feature_mean("mean",accel_gyro_data)
    df2 = feature_median("median",accel_gyro_data)
    df3 = feature_std("std",accel_gyro_data)
    df4 = feature_kurtosis("kurtosis",accel_gyro_data)
    df5 =feature_skewness("skewness",accel_gyro_data)
    df6 = feature_sum("sum",accel_gyro_data)
    df7 = feature_zero_cross("zero_cross_rate", accel_gyro_data)
    df8 = diff_sum("diff_sum", accel_gyro_data)
    df = [df1, df2, df3, df4, df5, df6,df7,df8]
    result_df = pd.concat(df, axis=1)

    return result_df
```

```python
def process_filter_code(df,column_name,Fs): # ' Accel_X'
    # 2) DETREND
    detrended_signal = signal.detrend(df[column_name])
    # 4) BAND PASS 1-5 Hz
    band_pass_filtered = butter_bandpass_filter(df[column_name], 1, 5, Fs, order=5)
    # 5) MOVIING AVERAGE  window_size = 5
    mov_average_filt = moving_average(band_pass_filtered, 5) #scipy.signal.medfilt(df_finger_tapping_2_5_index_norm[0], kernel_size=9)
    return mov_average_filt
```

```python
# Takes: An n-dimentional numpy array.
# Returns: yf,xf,AMP where yf is the scipi.fft() output, xf is the frequencies of the given fft, and AMP is the amplitude of the given fft. 
def my_fft(df,Fs):
    N = len(df)
    yf = scipy.fftpack.fft(df,n=N)
    # n, period of the signal.
    #print("my_fft() N//2: ",N//2)
    xf = scipy.fftpack.fftfreq(N,1/Fs)[:N//2] # get the frequency component Fs  = 40 Hz
    #print("my_fft() xf:",xf)
    #print("xf: ",xf)
    AMP= 2.0/N * np.abs(yf[0:N//2])
    #print("AMP: ",AMP)
    return yf,xf,AMP
```

```python
def all_axis_spectograms(fs,df):
    print("all_axis_spectograms(): Data frame shape: ",df.shape)
    # Compute the minimum length of all signals in the DataFrame
    min_signal_length = min(df.apply(len))
    # Calculate a reasonable value for nperseg
    nperseg = min_signal_length // 10  # You can adjust this factor as needed
    print("all_axis_spectograms(): min_signal_length: ===",min_signal_length)
    print("all_axis_spectograms(): nperseg: === ",nperseg)
    # Create a 2x3 grid of subplots
    fig, axs = plt.subplots(3, 3, figsize=(15, 10))
    mesh_list = []  # List to store the mesh objects
    # Iterate through each accelerometer axis
    for i, axis in enumerate([' Accel_X',' Accel_Y',' Accel_Z']):
        # Compute spectrogram for the current axis
        f, t, Sxx = signal.spectrogram(df[axis], fs=fs,nperseg=nperseg,noverlap = nperseg // 3) # nperseg=nperseg, noverlap=noverlap
        print("all_axis_spectograms(): Length of f: ",len(f),"all_axis_spectograms(): Length of t: ",len(t))
        # Welch Methods.
        f, Pxx_den = signal.welch(df[axis], fs, nperseg=nperseg,noverlap = nperseg // 3)
        # Plot spectrogram in the corresponding subplot
        row = i // 3
        col = i % 3
        mesh = axs[row, col].pcolormesh(t, f, 10 * np.log10(Sxx), shading='gouraud')
        axs[row, col].set_ylabel('Frequency [Hz]')
        axs[row, col].set_xlabel('Time [s]')
        axs[row, col].set_title(f'Spectrogram of {axis}')
        mesh_list.append(mesh)  # Append the mesh object to the lis
        # Add color bar
        cbar = fig.colorbar(mesh, ax=axs[row, col])
        cbar.set_label('Intensity (dB)')  # Set the label for the color bar
    axs[1, 0].plot(df[' Accel_X'],color="#171ad1", linestyle='--', label="RawData")
    axs[1, 1].plot(df[' Accel_Y'],color="#171ad1", linestyle='--', label="RawData")
    axs[1, 2].plot(df[' Accel_Z'],color="#171ad1", linestyle='--', label="RawData")
    axs[2, 0].semilogy(f, Pxx_den) 
    axs[2, 0].set_ylabel('PSD')
    axs[2, 0].set_xlabel('frequency [Hz]')
    axs[2, 0].set_ylim([0.5e-3, 1])
    axs[2, 1].semilogy(f, Pxx_den)
    axs[2, 1].set_ylabel('PSD')
    axs[2, 1].set_xlabel('frequency [Hz]')
    axs[2, 1].set_ylim([0.5e-3, 2])
    axs[2, 2].semilogy(f, Pxx_den) 
    axs[2, 1].set_ylim([0.5e-3, 2])
    axs[2, 2].set_ylabel('PSD')
    axs[2, 2].set_xlabel('frequency [Hz]')
    
    plt.tight_layout()
    plt.show()
```

```python
def activity_duration(df):
    timestamp_str1 = df["NewClock"][0]
    timestamp_str2 = df["NewClock"][df["NewClock"].shape[0]-1]
    
    # Convert string timestamps to datetime objects
    time_format = "%H:%M:%S"
    time1 = datetime.strptime(timestamp_str1, time_format)
    time2 = datetime.strptime(timestamp_str2, time_format)
    
    # Calculate the time difference
    time_difference = time2 - time1
    
    # Convert the time difference to seconds
    time_difference_seconds = time_difference.total_seconds()
    return time_difference_seconds
    
    
```

```python
# Reads each blobs of segments and calculates the time each blob ouccured. 
def read_all_blobs(folder_path):
    dfs = []
    file_list = os.listdir(folder_path)
    filename_list = []
    time_difference_seconds_list = []
    pid_list = []
    hand_list = []
    activity_list = []
    for filename in file_list:
        if filename.endswith('.csv'):
            substring1 = "Talking_to_peers"
            substring2 = "Talking_to_self"
            if substring1 in filename or substring2 in filename:
                print("Skipping Talking_to_peers")
                print("Skipping Talking_to_self")
                pass
            else:
                file_path = os.path.join(folder_path, filename)
                print("===========================================================================================================")
                print("filename: ",filename)
                df = pd.read_csv(file_path)
                time_difference_seconds = activity_duration(df)         
                pid = filename.split("_")[0]
                hand = filename.split("_")[1]
                activity = filename.split("_")[2]
                filename_list.append(filename)
                time_difference_seconds_list.append(time_difference_seconds)
                print("pid: ==== ",pid)
                pid_list.append(pid)
                hand_list.append(hand)
                activity_list.append(activity)
    df_duration = pd.DataFrame({'Filename': filename_list, 'Duration': time_difference_seconds_list, 'PID': pid_list,'hand_list': hand_list,'activity_id': activity_list,})
    return df_duration
```

```python
# Takes in an activity duration dataframe and outputs the duration of each behaviour plot for each participant.
def durations_plot_by_activity(duration_df_a1):
    adhd = [7, 9, 11, 14, 16, 24]
    fig = px.bar(duration_df_a1[duration_df_a1["newFeatureName"]!="NH"],x="PID_int", y="Duration", color_discrete_map=color_map,color="newFeatureName",
                 title=duration_df_a1["activity_id"].unique()[0],template="ggplot2", #category_orders={"PID": duration_df_a1["PID_int"].unique()}
                )
    # Set the tickmode of the x-axis to "array" and provide a list of all unique PID_int values
    fig.update_xaxes(tickmode="array", tickvals=duration_df_a1["PID_int"].unique())
    
    # Add annotation for missing values on the x-axis with specific y-axis tick values
    missing_values = set(set(list(duration_df["PID_int"].unique())) - set(list(duration_df_a1["PID_int"].unique())))
    print("missing_values",missing_values)
    
    # Add annotation for missing values and marked participant IDs below the x-axis
    missing_values = set(set(list(duration_df["PID_int"].unique())) - set(list(duration_df_a1["PID_int"].unique())))
    for value in missing_values.union(set(adhd)):
        color = "blue" if value in adhd else "black"  # Set color based on whether it's a marked participant ID
        fig.add_annotation(x=value, y=-0.1, text=str(value), showarrow=False, font=dict(color=color), yref="paper")

  
    return fig
```

```python
fig_a1 = durations_plot_by_activity(duration_df_a1)
fig_a1.show()
```
