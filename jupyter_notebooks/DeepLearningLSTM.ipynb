```python
def segments(df, time_steps, step, label_name, N_FEATURES):
    segments = []
    labels = []
    for i in range(0, len(df) - time_steps, step):
        xs = df[' Accel_X'].values[i:i+time_steps]
        ys = df[' Accel_Y'].values[i:i+time_steps]
        zs = df[' Accel_Z'].values[i:i+time_steps]
        g_xs = df[' Gyro_X'].values[i:i+time_steps]
        g_ys = df[' Gyro_Y'].values[i:i+time_steps]
        g_zs = df[' Gyro_Z'].values[i:i+time_steps]
        print("labels: ", mode(df["type"][i:i+time_steps]))  # Corrected indentation
        label = mode(df["type"][i:i+time_steps])
        print("Accelerometer ============ ")
        print("xs: ",xs,"\n len := \n",len(xs) ,"ys: ",ys,"\n len := \n",len(ys)  ,"\n len := \n","zs: ",zs,len(zs))
        print("Gryoscope ============ ")
        print("\n g_xs:",g_xs,"\n len := \n",len(g_xs) ,"\n g_ys: ",g_ys,"\n len := \n", len(g_ys) ,"\n len := \n","\n g_zs: ",g_zs,len(g_zs))
        segment = [xs, ys, zs,g_xs,g_ys,g_zs]
        segments.append(segment)
        print("segments():/ segment: ",type(segment),len(segment)) # segments():/ segment:  <class 'list'> 6
        labels.append(label)
    print("segments: ", np.asarray(segments).shape)# np.asarray(segments).shape = (112, 6, 56)  # Corrected indentation and removed extra space
    print("Segments shape:",-1, time_steps, N_FEATURES)
    print("time_steps: ",time_steps)
    reshaped_segments = np.asarray(segments, dtype=np.float32).reshape(-1, time_steps, N_FEATURES)
    print("reshaped_segments shape: ", reshaped_segments, reshaped_segments.shape)
    print("len of lables = ",len(labels),type(labels))
    labels = np.asarray(labels)
    # Get the count of each unique value
    counts = Counter(labels)
    print("Value counts of each unique label: ")
    # Print the counts
    for value, count in counts.items():
        print(f"{value}: {count}")
    return reshaped_segments, labels
```

```python
# Assuming df is your DataFrame. Splits the data frame into 80/20 % by time order.
def df_ordered_split(df):
    # Calculate the index for splitting
    split_index = int(0.8 * len(df))
    
    # Split the DataFrame
    train_df = df.iloc[:split_index]
    test_df = df.iloc[split_index:]
    
    # Optionally, reset index if needed
    train_df = train_df.reset_index(drop=True)
    test_df = test_df.reset_index(drop=True)
    return train_df,test_df

# Now train_df contains 80% of the original DataFrame, and test_df contains 20%.
```

```python
def resample_undersample(dataframe):
    y = dataframe["type"]
    X = dataframe.drop(columns=['type'])
    print("Number of values in each class: ",sorted(Counter(y).items()))

    rus = RandomUnderSampler(random_state=0)
    X_resampled, y_resampled = rus.fit_resample(X, y)
    concatenated_df = pd.concat([X_resampled, y_resampled], axis=1)

    # print("X_resampled: ",X_resampled)
    print("Resampled Number of values in each class: ", sorted(Counter(y_resampled).items()))
    return  concatenated_df
```

```python
def define_class_weights_navie(df):
    vals_dict = {}
    for i in df["type"]:
        if i in vals_dict.keys():
            vals_dict[i] += 1
        else:
            vals_dict[i] = 1
    total = sum(vals_dict.values())
    
    # Formula used - Naive method where
    # weight = 1 - (no. of samples present / total no. of samples)
    # So more the samples, lower the weight
    
    weight_dict = {k: (1 - (v / total)) for k, v in vals_dict.items()}
    print(weight_dict)
    return weight_dict
```

```python
### Code from tutorial - https://keras.io/examples/timeseries/timeseries_classification_from_scratch/
def conNet_timeseries(input_shape,num_classes):
    print("conNet_timeseries(): input_shape: ",input_shape)
    input_layer = keras.layers.Input(input_shape)

    conv1 = keras.layers.Conv1D(filters=64, kernel_size=3, padding="same")(input_layer)
    conv1 = keras.layers.BatchNormalization()(conv1)
    conv1 = keras.layers.ReLU()(conv1)

    conv2 = keras.layers.Conv1D(filters=64, kernel_size=3, padding="same")(conv1)
    conv2 = keras.layers.BatchNormalization()(conv2)
    conv2 = keras.layers.ReLU()(conv2)
    conv3 = keras.layers.Conv1D(filters=64, kernel_size=3, padding="same")(conv2)
    conv3 = keras.layers.BatchNormalization()(conv3)
    conv3 = keras.layers.ReLU()(conv3)

    gap = keras.layers.GlobalAveragePooling1D()(conv3)

    output_layer = keras.layers.Dense(1, activation="sigmoid")(gap)

    return keras.models.Model(inputs=input_layer, outputs=output_layer)
```

```python
######################## To plot the model metrics ########################
def plot_model_evaluvations(history,model,y_train,x_train):
    plt.figure(figsize=(6, 4))
    # plt.plot(history.history['sparse_categorical_accuracy'], 'r', label='sparse_categorical_accuracy of training data') # sparse_categorical_accuracy,.history['accuracy']
    plt.plot(history.history['accuracy'], 'b', label='accuracy of training data') # sparse_categorical_accuracy,.history['accuracy']
    # plt.plot(history.history['f1_score'], 'g', label='f1_score of training data') # sparse_categorical_accuracy,.history['accuracy']
    plt.plot(history.history['loss'], 'r--', label='Loss of training data')
    plt.title('Model Accuracy and Loss')
    plt.ylabel('Accuracy and Loss')
    plt.xlabel('Training Epoch')
    plt.ylim(0)
    plt.legend()
    plt.show()
    ################################## Model predictions Train #########################################################
    y_pred_train = model.predict(x_train)
    print("plot_model_evaluvations(history,model) y_pred_train: ",type(y_pred_train),y_pred_train)
    y_pred_train_binary = np.where(y_pred_train >= 0.5, 1, 0)
    # max_y_pred_train = np.argmax(y_pred_train, axis=1)
    print("plot_model_evaluvations(history,model) max_y_pred_train: ",type(y_pred_train_binary),y_pred_train_binary)
    clf_report = classification_report(y_train, y_pred_train_binary)
    conf_matrix = confusion_matrix(y_train, y_pred_train_binary)
    return plt,clf_report,conf_matrix
    
```

```python
def plot_confusion_matrix(cm, classes, title='Confusion matrix', cmap=plt.cm.Blues):
    """
    This function prints and plots the confusion matrix.
    """
    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)
    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], 'd'),
                 horizontalalignment="center",
                 color="orange") # if cm[i, j] > thresh else "orange"
    plt.tight_layout()
    plt.ylabel('True label')
    plt.xlabel('Predicted label')
    plt.show()
```
