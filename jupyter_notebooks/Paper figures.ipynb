```python
 
# Define a mapping function
def map_values(value):
    if pd.isna(value):
        return 0
    elif 'chair' in value:
        return 1
    elif 'Standing' in value:
        return 2
    elif "Manipulating" in value:
        return 3
    elif "Twirling" in value:
        return 4
    elif "Drumming" in value:
        return 5
    elif "Finger" in value:
        return 6
    elif "NoHyp" in value:
        return 0     
    else:
        return -1  # for values not matching any criteria

# Define a mapping function
def map_values_handedness(value):
    if pd.isna(value):
        return 0
    elif 'Left' in value:
        return "Left"
    elif 'Right' in value:
        return "Right"
    else:
        return -1  # for values not matching any criteria
 # Function to extract the time part from the timestamp
def extract_time(timestamp):
    return timestamp.split(':')[1:]
def get_new_int_timestamps(sorted_data):
    np_ts_array_list = []
    counter = 1
    for i in sorted_data:
        print("counter ===============",counter)
        print(i, ",",sorted_data[i])
        int_ts = np.full(sorted_data[i], counter)
        for i in int_ts:
            np_ts_array_list.append(i)
        print("int_ts: ",int_ts)
        counter +=1
    return np_ts_array_list
```

```python
def raw_plots_plotly(df):
    fig = make_subplots(rows=2, cols=3,
                        subplot_titles=('Accelerometer', 'Accel_y', 'Accel_Z',
                                        'Gryoscope', 'Gyro_Y', 'Gyro_Z')) 
    # Plotting Accelerometer data # 
    fig.add_trace(go.Scatter(x=df.index, y=df[' Accel_X'], mode='lines', name='Acc X',line=dict(color="#1e81b0")), row=1, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df['BehaviorCode'], mode='lines', name='Behavior Code',line=dict(color="#21130d")),row=1, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df[' Accel_Y'], mode='lines', name='Acc Y', line=dict(color="#e28743")), row=1, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df[' Accel_Z'], mode='lines', name='Acc Z',line=dict(color="#063970")), row=1, col=1)

    # Plotting Gyroscope data
    fig.add_trace(go.Scatter(x=df.index, y=df[' Gyro_X'], mode='lines', name='Gryo X',line=dict(color="#1e81b0")), row=2, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df['BehaviorCode'], mode='lines', name='Behavior Code',line=dict(color="#21130d") ),row=2, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df[' Gyro_Y'], mode='lines', name='Gryo Y',line=dict(color="#e28743")),row=2, col=1)
    fig.add_trace(go.Scatter(x=df.index, y=df[' Gyro_Z'], mode='lines', name='Gryo Z',line=dict(color="#063970")),row=2, col=1)
    fig.update_layout(template="simple_white")

    # Update layout
    fig.update_layout(height=600, width=1000, title_text="Filtered Signals",
                      title_font=dict(size=20), font=dict(size=10))
    return fig
```

```python
# calculate the MAX.
max_A = P13_Left[[' Accel_X', ' Accel_Y',' Accel_Z']].max(axis=1)
max_G = P13_Left[[' Gyro_X', ' Gyro_Y', ' Gyro_Z']].max(axis=1)

P13_Left["Acc_max"] = max_A
P13_Left["Gry_max"] = max_G

### Remove data when activity stopped.
P13_Left = P13_Left[P13_Left["Hyperactive_Restless"]!='-100']#["Hyperactive_Restless"].unique()
P13_Left = P13_Left[P13_Left["Hyperactive_Restless"]!=-100]
P13_Left['BehaviorCode'] = P13_Left['HyperactiveBehaviourType'].apply(map_values)

# Sort the dictionary by timestamp keys
sorted_data = dict(sorted(P13_Left["NewClock"].value_counts().to_dict().items(), key=lambda x: extract_time(x[0])))
np_ts_array_list = get_new_int_timestamps(sorted_data)

P13_Left["Int_TS"] = np_ts_array_list
```
