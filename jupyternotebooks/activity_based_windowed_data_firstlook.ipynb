```python
# 12864 + 7457+1520+1147 + 155 + 148

# Define a function to apply to each value in the 'Behaviour code' column
def label_behaviour(value):
    if value == 1:
        return 1
    elif value == 2:
        return 1
    elif value == 3:
        return 1
    elif value == 4:
        return 1
    elif value == 5:
        return 1
    elif value == 6:
        return 1
    elif value == 0:
        return 0
    else:
        return 0
```

```python

# Create a new column 'New Column' based on the conditions
df_activity_windowed_psd['type'] = df_activity_windowed_psd['BehaviorCode'].apply(label_behaviour)
```

```python
# Takes in an activity duration dataframe and outputs the duration of each behaviour plot for each participant.
def durations_plot_by_activity(df):

    fig = px.bar(df, x="PID_int", y="mean",
             color='FeatureName', barmode='group',
             height=400)
    # Set the tickmode of the x-axis to "array" and provide a list of all unique PID_int values
    fig.update_xaxes(tickmode="array", tickvals=df["PID_int"].unique())

  
    return fig
```
